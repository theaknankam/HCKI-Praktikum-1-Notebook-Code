```python
from groq import Groq
from google.colab import userdata

# Initialize the Groq client
api_key = userdata.get('GROQ_API_KEY')
client = Groq(api_key=api_key)

# Welcome message to interact with the bot
# Chatbot will run in infinite loop

print("Chatbot (Groq Streaming): Type 'quit', 'exit' or 'bye' to stop\n")
print("Welcome to the University of applied sciences of Cologne :)\n")

while True:
    user_input = input("You: ")

    if user_input.lower() in ["quit", "exit", "bye"]:
        print("\nChatbot: Goodbye!")
        break

    # Before the AI replies it will show chatbot: .....
    print("Chatbot: ", end="", flush=True)

    stream = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": user_input},
        ],
        stream=True,
    )

    for chunk in stream:
        content = chunk.choices[0].delta.content
        if content is not None:
            print(content, end="", flush=True)

    print()  # New line after response
```

```python
from groq import Groq
from gtts import gTTS
from IPython.display import Audio, display
import io


# Initialize Groq client
api_key = userdata.get('GROQ_API_KEY')
client = Groq(api_key=api_key)

# System prompt for TH Köln assistant
SYSTEM_PROMPT = """You are an enthusiastic and interactive study assistant for TH Köln (www.th-koeln.de).

TARGET GROUPS:
- Prospective students
- Current students (including first-year)
- Parents of students

🔴 CRITICAL RULES - YOU MUST FOLLOW THESE:
1. Be genuinely curious about their situation
2. You're having a CONVERSATION, not giving a lecture
3. You are a guide and suppose to help students finding their study programmex
4. You are here to support the study office not to replace it

EXAMPLE QUESTIONS TO ASK:
- "What program are you interested in?"
- "Are you a prospective or current student?"
- "What's your main concern right now?"
- "Would you like to know about deadlines, requirements, or course content?"
- "Is there a specific aspect I can help clarify?"

AREAS YOU SUPPORT:
- Study programs and academic advising
- Exams and regulations
- Study planning and organization
- Study abroad programs
- Administrative processes
- Graduation requirements

YOUR STYLE:
- Warm, friendly, conversational
- Keep responses brief (2-4 sentences with break between them to facilite the lisisbility)
- Always respond in English
- Show genuine interest in helping

"""

# Conversation history
conversation_history = [{"role": "system", "content": SYSTEM_PROMPT}]

print("=" * 70)
print("🎓 TH Köln Voice-Enabled Study Assistant 🔊")
print("=" * 70)
print("\nWelcome to the University of Applied Sciences of Cologne!")
print("I can help you with questions about studying at TH Köln.")
print("\nType 'quit', 'exit' or 'bye' to stop\n")
print("=" * 70 + "\n")


user_input = input("💬 You: ")

while True:

    if user_input.lower() in ["quit", "exit", "bye"]:
        print("\n🤖 Chatbot: Goodbye! Thank you for using TH Köln Study Assistant. Good luck with your studies!")

        # Say goodbye with voice
        tts = gTTS(text="Goodbye! Thank you for using TH Köln Study Assistant. Good luck with your studies!", lang='en')
        audio_fp = io.BytesIO()
        tts.write_to_fp(audio_fp)
        audio_fp.seek(0)
        display(Audio(audio_fp.read(), autoplay=False))
        break

    print("\n🤖 Chatbot: ", end="", flush=True)

    # Add user message to history
    conversation_history.append({"role": "user", "content": user_input})

    # Get response from Groq with conversation history
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=conversation_history,
        stream=False,
        temperature=0.9,
        max_tokens=1000,
    )

    bot_response = response.choices[0].message.content

    # Remove the reminder, add assistant response to history
    conversation_history.pop()  # Remove reminder
    conversation_history.append({"role": "assistant", "content": bot_response})


    # Convert response to speech
    try:
        tts = gTTS(text=bot_response, lang='en')
        audio_fp = io.BytesIO()
        tts.write_to_fp(audio_fp)
        audio_fp.seek(0)

        # Display audio player
        display(Audio(audio_fp.read(), autoplay=False))


    except Exception as e:
        print(f"⚠️ Audio error: {e}")
        print()

    print(bot_response)
    # Add empty line for spacing
    print()
    print()
    print()
    print()
    user_input = input("💬 You: ")
```
